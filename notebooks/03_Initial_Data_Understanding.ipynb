```python
# ============================================================
# C4.6 Initial Dataset Statistics and Structure Summary
# ============================================================

dataset_structure_summary = pd.DataFrame(index=merged_dataset.columns)

dataset_structure_summary["Data Type"] = merged_dataset.dtypes.astype(str)

dataset_structure_summary["Non-Null Count"] = merged_dataset.count()

dataset_structure_summary["Missing Values"] = merged_dataset.isnull().sum()

dataset_structure_summary["Unique Values"] = merged_dataset.nunique()

dataset_structure_summary["Memory Usage (MB)"] = (
    merged_dataset.memory_usage(deep=True).iloc[1:].values / (1024**2)
).round(2)

dataset_structure_summary
```

```python
# Save Initial Dataset Statistics and Structure Summary

dataset_structure_summary.to_csv(
    PROCESSED_DATA_DIR / "dataset_structure_summary.csv",
    index=False
)

print("Initial Dataset Statistics and Structure Summary saved successfully.")
```

```python
## C4.6 Initial Dataset Statistics and Structure Summary

The previous sections focused on understanding the engineering meaning and analytical role of each recorded variable. This section provides a structural overview of the merged operational dataset by summarising the characteristics of every variable.

Since the dataset has not yet undergone preprocessing, the objective is not to perform numerical statistical analysis but rather to examine its overall structure and quality.

For each variable, the following characteristics are summarised:

- Data type currently assigned by pandas.
- Number of non-null observations.
- Number of missing values.
- Number of unique recorded values.
- Approximate memory usage.

These statistics help assess the current state of the dataset and identify issues that will be addressed during the data cleaning stage.

It is expected that most operational variables are currently stored as `object` (or string-like) data types because the original datasets contain metadata rows describing measurement units and variable names. These structural characteristics will be corrected during Notebook 04 – Data Cleaning before any numerical statistical analysis is performed.
```

```python
### Overall Dataset Summary

The following summary provides an overview of the merged operational dataset, including its overall size, memory requirements, completeness, and readiness for subsequent preprocessing.
```

```python
# ============================================================
# Overall Dataset Summary
# ============================================================

total_memory_mb = (
    merged_dataset.memory_usage(deep=True).sum() / (1024 ** 2)
)

overall_summary = pd.DataFrame({

    "Metric":[

        "Total observations",

        "Total variables",

        "Operating-hour experiments",

        "Total missing values (NaN)",

        "Duplicate rows",

        "Metadata rows",

        "Current data type status",

        "Overall memory usage (MB)",

        "Dataset readiness"

    ],

    "Value":[

        f"{merged_dataset.shape[0]:,}",

        merged_dataset.shape[1],

        merged_dataset["operating_hour"].nunique(),

        merged_dataset.isnull().sum().sum(),

        merged_dataset.duplicated().sum(),

        "2 per operational dataset",

        "Most operational variables stored as object",

        round(total_memory_mb,2),

        "Requires data cleaning"

    ]

})

overall_summary
```

```python
# Save Overall Dataset Summary

overall_summary.to_csv(
    PROCESSED_DATA_DIR / "overall_summary.csv",
    index=False
)

print("Overall Dataset Summary saved successfully.")
```
