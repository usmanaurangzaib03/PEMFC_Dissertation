```python
## Define Project Paths

This section defines reusable folder paths for the project. Using paths instead of hard-coded file locations makes the notebook easier to maintain and reproduce.
```

```python
# ============================================================
# Define Project Paths
# ============================================================

# Current notebook location may be inside the notebooks folder.
# Therefore, we define the project root as the parent folder of the notebooks folder.
PROJECT_ROOT = Path.cwd().parent

# Define important project folders
DATA_DIR = PROJECT_ROOT / "data"
RAW_DATA_DIR = DATA_DIR / "raw"
PROCESSED_DATA_DIR = DATA_DIR / "processed"
FIGURES_DIR = PROJECT_ROOT / "figures"
RESULTS_DIR = PROJECT_ROOT / "results"

# Display paths
print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DATA_DIR)
print("Processed data folder:", PROCESSED_DATA_DIR)
print("Figures folder:", FIGURES_DIR)
print("Results folder:", RESULTS_DIR)
```

```python
## Locate Raw Dataset Files

This section identifies all CSV files stored in the raw data folder.
```

```python
# ============================================================
# Locate Raw Dataset Files
# ============================================================

# Find all CSV files in the raw data folder
csv_files = sorted(RAW_DATA_DIR.glob("*.csv"))

# Display number of files found
print(f"Number of CSV files found: {len(csv_files)}")

# Display file names
for file in csv_files:
    print(file.name)
```

```python
## Load the Dataset Files

This section loads all CSV files into Python as pandas DataFrames and stores them in a dictionary. Each dataset is accessed using its filename as the key.
```

```python
# ============================================================
# Load All CSV Files
# ============================================================

# Dictionary to store all datasets
datasets = {}

# Read each CSV file
for file in csv_files:
    datasets[file.stem] = pd.read_csv(file)

print(f"Successfully loaded {len(datasets)} datasets.")
```

```python
## Verify Successful Data Loading

After loading the files, the datasets are checked to confirm that each file was imported correctly. This includes checking the number of rows, columns, missing values, data types, and memory usage.
```

```python
# ============================================================
# Verify Successful Data Loading
# ============================================================

summary_data = []

for name, df in datasets.items():
    summary_data.append({
        "dataset_name": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_values": df.isna().sum().sum(),
        "duplicate_rows": df.duplicated().sum(),
        "memory_usage_MB": round(df.memory_usage(deep=True).sum() / (1024 ** 2), 2)
    })

dataset_summary = pd.DataFrame(summary_data)

dataset_summary
```

```python
# ============================================================
# Display Column Names for Each Dataset
# ============================================================

for name, df in datasets.items():
    print(f"\nDataset: {name}")
    print("-" * 80)
    print(list(df.columns))
```

```python
## C3.6.8 Dataset Consistency Verification

This section checks whether the operational datasets have a consistent structure. The verification includes checking column names, column order, number of columns, and data types across all operational files.
```

```python
# ============================================================
# C3.6.8 Dataset Consistency Verification
# ============================================================

# Separate operational datasets from polarization curve datasets
operational_datasets = {
    name: df for name, df in datasets.items()
    if "Polarization" not in name
}

polarization_datasets = {
    name: df for name, df in datasets.items()
    if "Polarization" in name
}

print(f"Operational datasets: {len(operational_datasets)}")
print(f"Polarization datasets: {len(polarization_datasets)}")
```

```python
# Use the first operational dataset as the reference structure
reference_name = list(operational_datasets.keys())[0]
reference_df = operational_datasets[reference_name]

reference_columns = list(reference_df.columns)
reference_dtypes = reference_df.dtypes.astype(str).to_dict()

consistency_results = []

for name, df in operational_datasets.items():
    consistency_results.append({
        "dataset_name": name,
        "same_number_of_columns": df.shape[1] == len(reference_columns),
        "same_column_names": set(df.columns) == set(reference_columns),
        "same_column_order": list(df.columns) == reference_columns,
        "same_data_types": df.dtypes.astype(str).to_dict() == reference_dtypes
    })

consistency_summary = pd.DataFrame(consistency_results)

consistency_summary
```

```python
## C3.7 Initial Dataset Summary

The following summary provides an overview of the imported datasets, including the number of datasets, observations, variables, data consistency, and overall readiness for further preprocessing.
```

```python
# ============================================================
# C3.7 Initial Dataset Summary
# ============================================================

summary = {
    "Total datasets": len(datasets),
    "Operational datasets": len(operational_datasets),
    "Polarization datasets": len(polarization_datasets),
    "Operational datasets structurally consistent": consistency_summary[
        ["same_number_of_columns",
         "same_column_names",
         "same_column_order",
         "same_data_types"]
    ].all().all(),
    "Operational datasets contain missing values":
        dataset_summary.loc[
            dataset_summary["dataset_name"].isin(operational_datasets.keys()),
            "missing_values"
        ].sum() > 0,
    "Operational datasets contain duplicate rows":
        dataset_summary.loc[
            dataset_summary["dataset_name"].isin(operational_datasets.keys()),
            "duplicate_rows"
        ].sum() > 0
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=["Item", "Result"]
)

summary_df
```

```python
## Polarization Dataset Verification

The polarization curve datasets are verified separately from the operational datasets because they represent a different experimental data structure. This section examines their dimensions, variables, data types, missing values, duplicate records, memory usage, and structural consistency.
```

```python
# ============================================================
# C3.9.2 Polarization Dataset Verification
# ============================================================

polarization_summary = []

for name, df in polarization_datasets.items():

    polarization_summary.append({
        "dataset_name": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_values": df.isna().sum().sum(),
        "duplicate_rows": df.duplicated().sum(),
        "memory_usage_MB": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 2
        )
    })

polarization_summary = pd.DataFrame(polarization_summary)

polarization_summary
```

```python
# ============================================================
# Compare Polarization Dataset Structures
# ============================================================

reference_polarization = list(polarization_datasets.values())[0]

comparison_results = []

for name, df in polarization_datasets.items():

    comparison_results.append({

        "dataset_name": name,

        "same_number_of_columns":
            df.shape[1] == reference_polarization.shape[1],

        "same_column_names":
            list(df.columns) == list(reference_polarization.columns),

        "same_data_types":
            df.dtypes.astype(str).tolist()
            ==
            reference_polarization.dtypes.astype(str).tolist()

    })

polarization_consistency = pd.DataFrame(comparison_results)

polarization_consistency
```

```python
## C3.9.3 Initial Polarization Dataset Observations

The verification process identified two polarization curve datasets within the GreenLight Innovation dataset.

Both datasets contain 48 observations; however, they differ in their overall structure. The `Polarization_curves` dataset contains 54 variables, whereas the `Polarization_curves_after_12h_rest` dataset contains 49 variables. This structural difference indicates that the datasets represent different experimental outputs and should not be treated as identical.

Both datasets contain missing values and three duplicate records. These observations suggest that additional preprocessing and validation will be required before the polarization datasets can be analysed.

A structural comparison further confirmed that the two polarization datasets differ in the number of variables, variable names, and data types. These differences are expected and do not indicate errors during the data loading process.

Unlike the operational time-series datasets, the polarization curve datasets represent fuel cell performance characterisation under specific operating conditions rather than continuous operational measurements. Consequently, they will be processed and analysed separately throughout the remainder of this research.

The original polarization datasets remain unchanged and are preserved within the `data/raw/` directory.
```
