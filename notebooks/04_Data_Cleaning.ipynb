```python
## C5.1 Structural Cleaning

The merged operational dataset currently contains structural elements that are not part of the experimental observations.

The objective of this section is to remove unnecessary information while preserving every genuine experimental measurement.

Structural cleaning includes:

- removing unnecessary variables;
- removing metadata rows;
- standardising the dataset structure.

These operations improve the quality of the dataset without modifying the underlying experimental data.
```

```python
### Remove Unnecessary Variables

The merged operational dataset contains the variable `Unnamed: 0`, which represents the exported row index from the original CSV files.

This variable does not describe any physical property of the PEM fuel cell and has no analytical value.

It is therefore removed before further preprocessing.
```

```python
# ============================================================
# Remove Unnecessary Variables
# ============================================================

cleaned_dataset = merged_dataset.drop(columns=["Unnamed: 0"])

print("'Unnamed: 0' removed successfully.")
```

```python
# ============================================================
# Verify Column Removal
# ============================================================

print("Number of variables after column removal:",
      cleaned_dataset.shape[1])

print("\nCurrent variables:\n")

print(list(cleaned_dataset.columns))
```

```python
### Verify Structural Integrity After Column Removal

Before continuing, the dataset structure is verified to ensure that only the unnecessary index column has been removed and that all experimental variables remain intact.
```

```python
# ============================================================
# Verify Structural Integrity After Column Removal
# ============================================================

print("Rows :", cleaned_dataset.shape[0])
print("Columns :", cleaned_dataset.shape[1])

print("\nMissing 'Unnamed: 0':",
      "Unnamed: 0" not in cleaned_dataset.columns)
```

```python
### Inspect Metadata Rows Before Removal

Before removing the metadata rows, it is important to confirm that they are present at the beginning of each operating-hour experiment.

The first operating-hour experiment (`50_h`) is inspected as a representative example. This allows us to visually distinguish the metadata rows from the genuine experimental observations and provides justification for their removal.

The inspection should demonstrate that:

- the first row contains measurement units;
- the second row contains variable descriptions;
- the third row represents the first genuine experimental observation.

Once this structure has been confirmed, the same cleaning procedure will be applied consistently to every operating-hour experiment.
```

```python
# ============================================================
# Inspect Metadata Rows Before Removal
# ============================================================

# Display first five rows of the 50-hour experiment

cleaned_dataset.loc[
    cleaned_dataset["operating_hour"] == 50
].head()
```

```python
### Remove Metadata Rows

Each operational dataset contains two metadata rows immediately following the header row. These rows describe the measurement units and variable names rather than representing experimental observations.

Using the `operating_hour` identifier, the first two rows of every operating-hour experiment are removed while preserving all genuine experimental measurements.

This approach ensures that:

- the same cleaning procedure is applied consistently to every experiment;
- no valid observations are removed;
- the integrity of the merged dataset is maintained.
```

```python
metadata_rows = (
    cleaned_dataset
    .groupby("operating_hour")
    .cumcount() < 2
)

cleaned_dataset = cleaned_dataset.loc[~metadata_rows].reset_index(drop=True)

print("Metadata rows removed successfully.")
print(cleaned_dataset.shape)
```

```python
### Verify Metadata Removal

After removing the metadata rows, the dataset is verified to ensure that:

- two rows have been removed from each operating-hour experiment;
- the first remaining row of each experiment is a genuine experimental observation;
- the total number of observations has decreased only by the expected amount.
```

```python
# ============================================================
# Verify Metadata Removal
# ============================================================

print("Dataset shape after metadata removal:")
print(cleaned_dataset.shape)

print("\n")

print("First five rows of the 50-hour experiment:")

cleaned_dataset.loc[
    cleaned_dataset["operating_hour"] == 50
].head()
```

```python
### Verify Structural Cleaning

Following the removal of the metadata rows, the integrity of each operating-hour experiment is verified.

The verification confirms that:

- every experiment remains present in the cleaned dataset;
- exactly two metadata rows have been removed from each experiment;
- the remaining observations correspond only to genuine experimental measurements;
- the `operating_hour` identifier has been preserved correctly.

This verification provides confidence that the cleaning procedure has been applied consistently across all operational datasets without affecting the recorded experimental observations.
```

```python
## C5.2 Data Type Conversion

After removing the metadata rows, the operational measurements can now be converted from string/object data types into numerical data types.

This conversion is necessary because statistical analysis, exploratory data analysis, feature engineering, and machine learning algorithms require numerical input variables.

The `operating_hour` column is retained as an integer identifier, while all sensor and operational measurement variables are converted to numeric format.
```

```python
# ============================================================
# C5.2 Data Type Conversion
# ============================================================

# Identify columns to convert to numeric
numeric_columns = [
    col for col in cleaned_dataset.columns
    if col != "operating_hour"
]

# Convert operational variables to numeric
for col in numeric_columns:
    cleaned_dataset[col] = pd.to_numeric(
        cleaned_dataset[col],
        errors="coerce"
    )

print("Data type conversion completed successfully.")
```

```python
# ============================================================
# Verify Data Types After Conversion
# ============================================================

cleaned_dataset.dtypes
```

```python
### Verify Data Type Conversion

After converting the operational variables to numerical data types, it is important to verify that the conversion was successful.

Because the conversion was performed using `errors='coerce'`, any values that could not be interpreted as numerical measurements would have been converted to missing values (`NaN`).

The following verification confirms that:

- all operational variables have been converted successfully;
- no unexpected missing values have been introduced during the conversion;
- the dataset remains suitable for subsequent analysis.
```

```python
# ============================================================
# Verify Missing Values After Data Type Conversion
# ============================================================

missing_after_conversion = (
    cleaned_dataset
    .isnull()
    .sum()
    .reset_index()
)

missing_after_conversion.columns = [
    "Variable",
    "Missing Values"
]

missing_after_conversion
```

```python
# ============================================================
# Verify Successful Numeric Conversion
# ============================================================

if cleaned_dataset.isnull().sum().sum() == 0:
    print("✓ Data type conversion completed successfully.")
    print("✓ No missing values were introduced.")
else:
    print("⚠ Missing values detected after conversion.")
```
