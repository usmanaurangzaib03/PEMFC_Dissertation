```python
## 8.3.1 Current Behaviour Profile

### Purpose

This section investigates how the electrical current behaves throughout the PEM fuel cell durability experiment.

Current represents the electrical load applied to the fuel cell. Because the current values change repeatedly according to the experimental operating schedule, plotting the complete high-resolution signal across all observations would produce a dense and difficult-to-interpret figure.

To improve clarity, the current behaviour profile will therefore be examined using two complementary views:

- a long-term stage-level view to compare current behaviour across the operating-hour experiments;
- a representative short time window to reveal the detailed loading pattern within selected experiments.

### Analysis Approach

The analysis will:

- calculate stage-level current behaviour across all operating-hour experiments;
- compare the central current level between durability stages;
- examine the current operating range;
- select representative early, middle, and late durability stages;
- extract a limited continuous time window from each selected stage;
- plot these windows separately to make repeated loading cycles and transitions visible.

A short continuous window is used instead of plotting the complete experiment because each stage contains approximately 179,000 observations. Displaying the entire signal would compress rapid current changes into an unreadable set of vertical lines. The selected-window approach preserves the original chronological sequence while providing a clearer view of the operating pattern.

### Expected Outputs

- Stage-level current behaviour summary.
- Long-term current comparison across durability stages.
- Representative current windows from early, middle, and late experiments.
- Clear visual evidence of loading cycles, transitions, operating range, and behavioural consistency.
```

```python
## Overall Subsystem Behaviour Analysis

Following the individual behaviour profiling of each operational variable, this section extends the analysis to the subsystem level by collectively examining variables belonging to the same engineering subsystem. The objective is to evaluate how the electrical, pressure, temperature, and reactant flow subsystems behave throughout the durability experiment, providing a holistic understanding of system operation that cannot be obtained from individual variable analyses alone.

For each subsystem, three complementary analyses are performed. First, stage-level normalised behaviour is used to compare the long-term temporal evolution and behavioural trends of the subsystem variables on a common scale. Second, stage-level variability analysis is conducted using statistical measures, including the median, quartiles, and interquartile range (IQR), to assess behavioural stability and variability throughout the durability experiment. Finally, representative operating behaviour is analysed using selected operating windows to investigate short-term operating characteristics, behavioural fluctuations, transitions, and the repeatability of subsystem operation under identical operating conditions.

Together, these analyses provide a comprehensive assessment of subsystem behaviour by evaluating temporal evolution, behavioural trends, stability, variability, fluctuations, behavioural transitions, and operating behaviour. The findings establish a subsystem-level understanding of PEM fuel cell operation and provide a robust foundation for the subsequent analyses of variable relationships, feature engineering, and machine learning-based degradation prediction.
```
