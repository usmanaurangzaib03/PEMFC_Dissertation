```python
## C7.4 Statistical Outlier Investigation

Outliers are observations that differ substantially from the majority of the recorded measurements.

In engineering datasets, outliers are not necessarily errors. They may represent unusual operating conditions, transient system behaviour, degradation effects, sensor noise, or genuine physical phenomena.

The objective of this section is to identify statistically unusual observations using two complementary approaches:

- Interquartile Range (IQR)
- Z-score Analysis

The identified observations will be documented for further interpretation rather than removed from the dataset.
```

```python
### Interquartile Range (IQR) Investigation

The Interquartile Range method is used to identify observations that fall substantially outside the central 50% of the recorded measurements.

For each operational variable, lower and upper thresholds are calculated using the 1.5 × IQR rule, and the number of observations outside these limits is recorded.
```

```python
# ============================================================
# Interquartile Range (IQR) Investigation
# ============================================================

numerical_columns = cleaned_dataset.select_dtypes(
    include="number"
).columns.tolist()

# Remove identifier
numerical_columns.remove("operating_hour")

iqr_results = []

for column in numerical_columns:

    Q1 = cleaned_dataset[column].quantile(0.25)
    Q3 = cleaned_dataset[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (cleaned_dataset[column] < lower) |
        (cleaned_dataset[column] > upper)
    ).sum()

    iqr_results.append({

        "Variable": column,
        "Q1": round(Q1,4),
        "Q3": round(Q3,4),
        "Lower Bound": round(lower,4),
        "Upper Bound": round(upper,4),
        "Outliers": outliers,
        "Percentage (%)": round(
            outliers / len(cleaned_dataset) * 100,
            4
        )

    })

iqr_summary = pd.DataFrame(iqr_results)

iqr_summary
```

```python
### Z-score Investigation

The Z-score method identifies observations located more than three standard deviations from the mean.

This provides a complementary assessment of statistical outliers and allows comparison with the Interquartile Range method.
```

```python
# ============================================================
# Z-score Investigation
# ============================================================

zscore_results = []

for column in numerical_columns:

    mean = cleaned_dataset[column].mean()
    std = cleaned_dataset[column].std()

    z_scores = (
        cleaned_dataset[column] - mean
    ) / std

    outliers = (
        np.abs(z_scores) > 3
    ).sum()

    zscore_results.append({

        "Variable": column,
        "Mean": round(mean,4),
        "Std Dev": round(std,4),
        "Outliers": outliers,
        "Percentage (%)": round(
            outliers / len(cleaned_dataset) * 100,
            4
        )

    })

zscore_summary = pd.DataFrame(zscore_results)

zscore_summary
```

```python
### Comparison of Outlier Detection Methods

The number of statistical outliers identified by the Interquartile Range and Z-score methods are compared for each operational variable.

This comparison highlights variables that consistently exhibit unusual behaviour across different statistical detection techniques.
```

```python
# ============================================================
# Compare IQR and Z-score Results
# ============================================================

comparison = pd.DataFrame({

    "Variable": numerical_columns,

    "IQR Outliers":
        iqr_summary["Outliers"],

    "IQR Percentage (%)":
        iqr_summary["Percentage (%)"],

    "Z-score Outliers":
        zscore_summary["Outliers"],

    "Z-score Percentage (%)":
        zscore_summary["Percentage (%)"]

})

comparison
```

```python
### Outlier Investigation Summary

The statistical outlier investigation summarises the observations identified by both detection methods.

These observations are retained within the dataset because they may represent genuine engineering behaviour rather than erroneous measurements.
```

```python
# ============================================================
# Outlier Investigation Summary
# ============================================================

outlier_summary = pd.DataFrame({

    "Metric":[

        "Variables assessed",

        "IQR method",

        "Z-score method",

        "Dataset readiness"

    ],

    "Value":[

        len(numerical_columns),

        "Completed",

        "Completed",

        "Ready for Sensor Spike Investigation (C7.5)"

    ]

})

outlier_summary
```

```python
## C7.4.5 Research Insight

### Observation

The statistical outlier investigation identified substantial differences between the Interquartile Range (IQR) and Z-score detection methods. The IQR method consistently identified a larger number of outliers than the Z-score method, particularly for pressure, dewpoint temperature, and reactant flow variables. In contrast, voltage and power contained very few statistically unusual observations.

### Why is this important?

Different statistical outlier detection methods respond differently to the underlying distribution of the data. The larger number of IQR outliers suggests that several operational variables exhibit skewed or non-normal distributions rather than measurement errors. Using multiple detection methods provides a more comprehensive understanding of the dataset and reduces the risk of incorrectly classifying valid observations as anomalies.

### Engineering Interpretation

The large number of statistical outliers observed in pressure, humidity (dewpoint), and reactant flow variables is consistent with the dynamic operating conditions of PEM fuel cell durability experiments. These variables naturally fluctuate in response to changes in reactant supply, water management, thermal regulation, and degradation processes. Conversely, the voltage variable exhibits remarkably few outliers, indicating relatively stable electrochemical behaviour throughout the experiments despite gradual performance degradation.

### Dissertation Interpretation

The statistical outlier investigation demonstrates that many unusual observations are likely to represent genuine operational behaviour rather than erroneous measurements. Consequently, statistical outliers should not be removed automatically. Instead, they should be retained for further investigation during exploratory data analysis and degradation behaviour analysis, where their physical significance can be evaluated within the operational context of the PEM fuel cell.

### Decision

No statistical outliers will be removed during this stage of the preprocessing workflow. All identified observations will be preserved for subsequent exploratory data analysis, where they will be interpreted in conjunction with operating conditions and degradation trends.
```

```python
## C7.5 Sensor Spike Investigation

Sensor spikes refer to sudden changes between consecutive measurements recorded during the operational experiments.

Unlike statistical outliers, which are identified from the overall distribution of the data, sensor spikes are detected by examining changes between successive observations in the time series.

This investigation helps identify abrupt measurement changes that may represent transient operating conditions, sensor artefacts, or genuine electrochemical behaviour.
```

```python
### First-Order Difference Calculation

The first-order difference measures the absolute change between consecutive observations for each operational variable.

Large first-order differences indicate sudden changes in the recorded measurements and are used as the basis for identifying potential sensor spikes.
```

```python
# ============================================================
# Calculate First-Order Differences
# ============================================================

first_difference = cleaned_dataset.copy()

for column in numerical_columns:
    first_difference[column] = (
        cleaned_dataset
        .groupby("operating_hour")[column]
        .diff()
        .abs()
    )

print("First-order differences calculated successfully.")
```

```python
### Largest Sensor Changes

The maximum first-order difference is calculated for each operational variable.

This identifies the largest observed change between consecutive measurements during the experiments.
```

```python
# ============================================================
# Largest Sensor Changes
# ============================================================

largest_changes = pd.DataFrame({

    "Variable": numerical_columns,

    "Maximum Change": [

        first_difference[column].max()

        for column in numerical_columns

    ]

})

largest_changes = largest_changes.sort_values(
    by="Maximum Change",
    ascending=False
)

largest_changes
```

```python
### Potential Sensor Spike Candidates

Potential sensor spikes are identified using a variable-specific threshold based on the distribution of first-order differences.

For each variable, observations exceeding the mean plus three standard deviations of the first-order differences are identified as potential spike candidates.
```

```python
# ============================================================
# Identify Potential Sensor Spikes
# ============================================================

spike_results = []

for column in numerical_columns:

    threshold = (
        first_difference[column].mean() +
        3 * first_difference[column].std()
    )

    spikes = (
        first_difference[column] > threshold
    ).sum()

    spike_results.append({

        "Variable": column,
        "Threshold": round(threshold, 4),
        "Spike Candidates": spikes,
        "Percentage (%)": round(
            spikes / len(cleaned_dataset) * 100,
            4
        )

    })

spike_summary = pd.DataFrame(spike_results)

spike_summary
```

```python
### Sensor Spike Summary

The sensor spike investigation summarises the occurrence of sudden measurement changes across the operational variables.

Potential spike candidates are documented for further interpretation during exploratory data analysis rather than being treated as erroneous observations.
```

```python
sensor_spike_summary = pd.DataFrame({

    "Metric":[

        "Variables investigated",

        "Spike detection method",

        "Total spike detections",

        "Variables containing spike candidates",

        "Dataset readiness"

    ],

    "Value":[

        len(numerical_columns),

        "First-order difference (Mean + 3×SD)",

        f"{spike_summary['Spike Candidates'].sum():,}",

        (spike_summary["Spike Candidates"] > 0).sum(),

        "Ready for Time-Series Integrity Investigation (C7.6)"

    ]

})

sensor_spike_summary
```

```python
## C7.5.5 Research Insight

### Observation

The sensor spike investigation identified abrupt changes across all operational variables, with the largest first-order differences occurring in pressure, current, and power measurements. Every operational variable contained some spike candidates, although the proportion of detected spikes remained relatively small for each individual variable.

### Why is this important?

Sensor spike analysis complements statistical outlier detection by examining the temporal behaviour of the measurements rather than their overall distribution. Sudden changes between consecutive observations may indicate transient operating conditions, control system responses, measurement artefacts, or genuine electrochemical behaviour that cannot be identified using conventional statistical methods alone.

### Engineering Interpretation

The larger number of spike candidates observed in pressure, current, power, and reactant flow variables is consistent with the dynamic nature of PEM fuel cell operation. Rapid adjustments in reactant supply, electrical loading, and control system responses can produce abrupt yet physically meaningful changes in these variables. In contrast, voltage and temperature variables exhibit comparatively smoother temporal behaviour, reflecting the slower electrochemical and thermal dynamics of the fuel cell.

### Dissertation Interpretation

The sensor spike investigation demonstrates that abrupt measurement changes are a natural characteristic of the operational dataset rather than clear evidence of poor data quality. These observations should be retained because they may contain valuable information regarding transient operating conditions and degradation behaviour. Their significance will be examined further during exploratory data analysis and degradation behaviour analysis.

### Decision

No sensor spike candidates will be removed during preprocessing. All observations will be preserved for subsequent exploratory analysis, where their engineering significance will be evaluated within the context of PEM fuel cell operation and degradation.
```

```python
## C7.5.5 Targeted Temporal Inspection

The sensor spike investigation identified observations with unusually large
changes between consecutive measurements. However, an abrupt change does not
necessarily indicate a sensor error because the PEMFC was operated under a
dynamic FC-DLC current profile.

This section investigates the temporal context surrounding the most substantial
spike events. Major events are located within their corresponding
`operating_hour` experiment and examined together with physically related
operational variables.

The objective is to distinguish between:

- controlled FC-DLC load transitions;
- normal controller or system transients;
- coordinated physical responses across multiple variables;
- isolated observations that may require further data-quality investigation.

No observations are removed during this investigation.
```

```python
### C7.5.5.1 Select Variables for Targeted Inspection

Targeted inspection is focused on variables showing particularly large
maximum consecutive changes.

The initial inspection focuses on:

- `pressure_anode_inlet`;
- `pressure_anode_outlet`;
- `temp_anode_outlet`;
- `temp_cathode_outlet`.

These variables were selected because their maximum one-step changes were
substantial relative to their normal operating ranges.

Electrical, pressure, temperature, and reactant-flow variables will also be
examined around each event to determine whether the detected change occurred
as part of a coordinated system transition or as an isolated measurement.
```

```python
# ============================================================
# Select Variables for Targeted Temporal Inspection
# ============================================================

target_spike_variables = [

    "pressure_anode_inlet",
    "pressure_anode_outlet",
    "temp_anode_outlet",
    "temp_cathode_outlet"

]

context_variables = [

    "current",
    "voltage",
    "power",

    "pressure_anode_inlet",
    "pressure_anode_outlet",
    "pressure_cathode_inlet",
    "pressure_cathode_outlet",

    "temp_anode_endplate",
    "temp_anode_inlet",
    "temp_anode_outlet",
    "temp_cathode_inlet",
    "temp_cathode_outlet",

    "total_anode_stack_flow",
    "total_cathode_stack_flow"

]

print("Target spike variables:")

for variable in target_spike_variables:
    print(f"- {variable}")

print(
    f"\nNumber of context variables: "
    f"{len(context_variables)}"
)
```

```python
### C7.5.5.2 Locate Maximum-Change Events

The maximum consecutive change for each selected variable is located within
the durability dataset.

For each event, the corresponding:

- `operating_hour`;
- timestamp;
- previous value;
- current value;
- and absolute consecutive change

are recorded.

This provides the exact temporal location required for subsequent
contextual inspection.
```

```python
# ============================================================
# Locate Maximum-Change Events
# ============================================================

maximum_event_records = []

for variable in target_spike_variables:

    # Index of the largest absolute first-order difference
    event_index = first_difference[variable].idxmax()

    # Current row
    current_row = cleaned_dataset.loc[event_index]

    # Find the previous observation within the same operating-hour group
    group_indices = cleaned_dataset[
        cleaned_dataset["operating_hour"]
        == current_row["operating_hour"]
    ].index

    event_position = group_indices.get_loc(event_index)

    if event_position > 0:

        previous_index = group_indices[event_position - 1]

        previous_value = cleaned_dataset.loc[
            previous_index,
            variable
        ]

    else:

        previous_index = None
        previous_value = np.nan

    current_value = current_row[variable]

    maximum_change = first_difference.loc[
        event_index,
        variable
    ]

    maximum_event_records.append({

        "Variable": variable,
        "Operating Hour": current_row["operating_hour"],
        "Event Index": event_index,
        "Previous Index": previous_index,
        "Time": current_row["time"],
        "Previous Value": previous_value,
        "Current Value": current_value,
        "Absolute Change": maximum_change

    })


maximum_events_df = pd.DataFrame(
    maximum_event_records
)

display(maximum_events_df)
```

```python
### C7.5.5.3 Extract Temporal Windows Around Maximum-Change Events

Temporal windows are extracted around each identified maximum-change event to
examine the behaviour immediately before and after the abrupt transition.

A window of 30 observations before and after each event is used. Since the
operational data were recorded at approximately 1 Hz, this represents roughly
30 seconds of behaviour on either side of the event.

The extraction is restricted to the corresponding `operating_hour` to prevent
observations from different durability stages from being combined.

These windows provide the temporal context required to determine whether each
event represents an isolated measurement anomaly or a coordinated PEMFC
operating-system transition.
```

```python
# ============================================================
# C7.5.5.3 — Extract Temporal Windows Around Maximum Events
# ============================================================

window_size = 30

temporal_windows = {}

for _, event in maximum_events_df.iterrows():

    variable = event["Variable"]
    operating_hour = event["Operating Hour"]
    event_index = int(event["Event Index"])

    # Select only observations from the same durability stage
    stage_data = cleaned_dataset[
        cleaned_dataset["operating_hour"] == operating_hour
    ]

    stage_indices = stage_data.index

    # Find event position within the durability stage
    event_position = stage_indices.get_loc(event_index)

    # Define safe window boundaries
    start_position = max(
        0,
        event_position - window_size
    )

    end_position = min(
        len(stage_data),
        event_position + window_size + 1
    )

    # Extract event window
    event_window = stage_data.iloc[
        start_position:end_position
    ].copy()

    # Mark the event observation
    event_window["Event"] = (
        event_window.index == event_index
    )

    # Store the window
    temporal_windows[variable] = event_window


print(
    f"Temporal windows extracted successfully "
    f"for {len(temporal_windows)} variables."
)

print(
    f"Window size: ±{window_size} observations"
)
```

```python
### C7.5.5.4 Inspect Cross-Variable Behaviour Around Major Events

The maximum-change events are examined together with related electrical,
pressure, temperature, and reactant-flow measurements.

Events occurring at the same `operating_hour` and timestamp are treated as
a single system event. A focused numerical window is displayed around each
unique event to determine whether abrupt changes occur simultaneously across
multiple PEMFC variables.

Coordinated changes across physically related variables may indicate a
controlled operating transition or system response, whereas an isolated
change may indicate a potential measurement anomaly.
```

```python
# ============================================================
# C7.5.5.4 — Inspect Cross-Variable Behaviour Around Major Events
# ============================================================

# Identify unique system events
unique_events_df = (
    maximum_events_df[
        [
            "Operating Hour",
            "Event Index",
            "Time"
        ]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

print(
    f"Number of unique major events: "
    f"{len(unique_events_df)}"
)

display(unique_events_df)


# ============================================================
# Variables Used for Cross-Variable Inspection
# ============================================================

inspection_variables = [

    "time",
    "current",
    "voltage",
    "power",

    "pressure_anode_inlet",
    "pressure_anode_outlet",
    "pressure_cathode_inlet",
    "pressure_cathode_outlet",

    "temp_anode_endplate",
    "temp_anode_inlet",
    "temp_anode_outlet",
    "temp_cathode_inlet",
    "temp_cathode_outlet",

    "total_anode_stack_flow",
    "total_cathode_stack_flow"

]


# ============================================================
# Display Focused Context Around Each Unique Event
# ============================================================

inspection_radius = 5

event_contexts = {}

for event_number, event in unique_events_df.iterrows():

    operating_hour = event["Operating Hour"]
    event_index = int(event["Event Index"])
    event_time = event["Time"]

    # Select corresponding durability stage
    stage_data = cleaned_dataset[
        cleaned_dataset["operating_hour"] == operating_hour
    ]

    stage_indices = stage_data.index

    # Find event position within stage
    event_position = stage_indices.get_loc(event_index)

    # Define safe inspection boundaries
    start_position = max(
        0,
        event_position - inspection_radius
    )

    end_position = min(
        len(stage_data),
        event_position + inspection_radius + 1
    )

    # Extract focused event context
    event_context = stage_data.iloc[
        start_position:end_position
    ][inspection_variables].copy()

    # Mark event observation
    event_context.insert(
        0,
        "Event",
        event_context.index == event_index
    )

    # Store for later use
    event_contexts[
        f"Event_{event_number + 1}"
    ] = event_context

    print(
        "\n"
        + "=" * 70
    )

    print(
        f"EVENT {event_number + 1}"
    )

    print(
        f"Operating Hour : {operating_hour}"
    )

    print(
        f"Event Time     : {event_time:.2f} s"
    )

    print(
        f"Event Index    : {event_index}"
    )

    print(
        "=" * 70
    )

    display(
        event_context.round(4)
    )
```

```python
# ============================================================
# Chronological Ordering Verification
# ============================================================

chronological_check = (

    cleaned_dataset

    .groupby("operating_hour")["time"]

    .apply(lambda x: x.is_monotonic_increasing)

    .reset_index(name="Chronological Order")

)

chronological_check
```

```python
### Sampling Interval Investigation

The sampling interval between consecutive observations is calculated for each operating-hour experiment.

Consistent sampling intervals are essential for reliable time-series analysis and sequential machine learning.
```

```python
# ============================================================
# Sampling Interval Investigation
# ============================================================

sampling_results = []

for hour, df in cleaned_dataset.groupby("operating_hour"):

    intervals = df["time"].diff().dropna()

    sampling_results.append({

        "Operating Hour": hour,

        "Mean Interval": intervals.mean(),

        "Minimum Interval": intervals.min(),

        "Maximum Interval": intervals.max(),

        "Standard Deviation": intervals.std()

    })

sampling_summary = pd.DataFrame(sampling_results)

sampling_summary
```

```python
### Time-Gap Investigation

Large time gaps between consecutive observations may indicate interruptions in data acquisition or missing measurements.

This investigation identifies unusually large sampling intervals within each operating-hour experiment.
```

```python
# ============================================================
# Time-Gap Investigation
# ============================================================

gap_results = []

for hour, df in cleaned_dataset.groupby("operating_hour"):

    intervals = df["time"].diff().dropna()

    expected_interval = intervals.median()

    large_gaps = (

        intervals >

        expected_interval * 2

    ).sum()

    gap_results.append({

        "Operating Hour": hour,

        "Large Time Gaps": large_gaps

    })

gap_summary = pd.DataFrame(gap_results)

gap_summary
```

```python
### Operating-Hour Experiment Integrity

The first and last timestamps of each operating-hour experiment are examined to verify that each experiment forms an independent and complete time series.
```

```python
# ============================================================
# Operating-Hour Experiment Integrity
# ============================================================

experiment_integrity = (

    cleaned_dataset

    .groupby("operating_hour")

    .agg(

        Observations=("time","count"),

        Start_Time=("time","min"),

        End_Time=("time","max")

    )

    .reset_index()

)

experiment_integrity["Duration (s)"] = (
    experiment_integrity["End_Time"] -
    experiment_integrity["Start_Time"]
).round(3)

# NEW
experiment_integrity["Duration (hours)"] = (
    experiment_integrity["Duration (s)"] / 3600
).round(2)

experiment_integrity
```
